# Exercise 3 — Self-Reflection Prompt for Improving Output

**Tool used:** Prompts tested in ChatGPT / OpenAI Playground; captured prompts and responses reproduced here.

**Goal:** Have the AI generate a summary, then critique its own output against explicit criteria, and revise it — demonstrating measurable improvement.

## Source text to summarize

*(Short original passage used for this exercise.)*

> Cloud computing has changed how small businesses operate. Instead of buying and maintaining
> physical servers, a small business can rent computing power, storage, and software from a
> provider and pay only for what it uses. This lowers the upfront cost of starting or scaling a
> business, since there is no need for a large IT budget or a dedicated technical staff. It also
> makes it easier for a small team to access tools that used to be available only to large
> companies, such as advanced data analytics or enterprise-grade security. On the other hand,
> relying on a third-party provider introduces new risks: if the provider has an outage, the
> business's own operations may stop working too, and switching providers later can be costly
> and technically difficult. Overall, cloud computing has lowered the barrier to entry for small
> businesses while introducing a new kind of dependency they need to manage.

## Step 1 — Generate the initial summary

**Prompt:**
```
Summarize the following article in 3 bullet points, focusing on its impact on small
businesses.

[article text above]
```

````python
original_summary_prompt = """Summarize the following article in 3 bullet points, focusing on its impact on small businesses.

[article text above]"""

# --- Captured initial ("before") output from the model ---
original_summary = """- Cloud computing lets businesses rent computing power and storage instead of buying physical servers, which is a major shift in how IT infrastructure works today.
- Small businesses can access advanced tools like analytics and security that used to be reserved for large companies with big budgets.
- There are also risks: outages at the provider can disrupt operations, switching providers is hard, and cloud adoption overall is a growing trend across every industry."""

print("PROMPT SENT:\n" + original_summary_prompt)
print("\nAI OUTPUT (BEFORE):\n" + original_summary)
````

## Step 2 — Self-reflection (critique the Step 1 output against explicit criteria)

**Requirements used for critique:**
- Exactly 3 bullet points
- Each bullet under 20 words
- Each bullet focused specifically on small-business impact (not general industry commentary)
- No filler or generic statements

**Prompt:**
```
Here is the summary you just generated:
[paste Step 1 output]

Critique your own summary against these criteria:
- Is it exactly 3 bullet points?
- Is each bullet under 20 words?
- Does each bullet focus specifically on small-business impact, not general commentary?
- Does it avoid filler language?

State clearly whether it passes each criterion. If it fails any, rewrite the summary to
fix the issues and briefly explain what changed.
```

````python
self_reflection_prompt = f"""Here is the summary you just generated:
{original_summary}

Critique your own summary against these criteria:
- Is it exactly 3 bullet points?
- Is each bullet under 20 words?
- Does each bullet focus specifically on small-business impact, not general commentary?
- Does it avoid filler language?

State clearly whether it passes each criterion. If it fails any, rewrite the summary to
fix the issues and briefly explain what changed."""

# --- Captured self-critique + revised ("after") output from the model ---
self_critique = """Checking against the criteria:
- Exactly 3 bullet points: PASS
- Each bullet under 20 words: FAIL (bullet 1 is 24 words, bullet 3 is 27 words)
- Focused on small-business impact specifically: FAIL (bullet 1 reads as general IT commentary; bullet 3 drifts into "growing trend across every industry," which is not small-business-specific)
- Avoids filler language: FAIL ("major shift in how IT infrastructure works today" and "growing trend across every industry" are filler that don't add small-business-specific information)

Revising to fix all three issues."""

revised_summary = """- Renting computing power instead of buying servers lowers the upfront cost for small businesses to start or scale.
- Small teams gain access to enterprise-grade tools, like analytics and security, once out of reach for their budgets.
- Provider outages or difficult provider switches can disrupt a small business's operations and create new dependency risks."""

print("PROMPT SENT:\n" + self_reflection_prompt)
print("\nAI SELF-CRITIQUE:\n" + self_critique)
print("\nAI OUTPUT (AFTER / REVISED):\n" + revised_summary)
````

## Before vs. After comparison

````python
def word_count(bullet):
    return len(bullet.lstrip("- ").split())

print("BEFORE:")
for b in original_summary.split("\n"):
    print(f"  ({word_count(b)} words) {b}")

print("\nAFTER:")
for b in revised_summary.split("\n"):
    print(f"  ({word_count(b)} words) {b}")

before_over_limit = sum(1 for b in original_summary.split("\n") if word_count(b) > 20)
after_over_limit = sum(1 for b in revised_summary.split("\n") if word_count(b) > 20)
print(f"\nBullets over the 20-word limit -> before: {before_over_limit}, after: {after_over_limit}")
````

## Result

The self-reflection step caught two concrete, checkable problems in the first draft — two bullets exceeding the 20-word limit, and language that drifted away from small-business-specific impact into generic industry commentary. The revised summary fixes both: every bullet is now under 20 words and stays focused specifically on how the effect applies to a small business (lower upfront cost, access to enterprise tools, dependency risk), satisfying all four criteria used for the critique.
