# Exercise 1 — Prompt Chaining for a Customer Support AI

**Tool used:** Prompts were designed and tested in the OpenAI Playground (GPT model) / ChatGPT interface. This notebook reproduces the exact prompts used at each step and the captured AI responses, so the full chain and its outputs are visible in one place for submission.

**Goal:** Build a multi-step prompt chain where the output of one prompt becomes the input to the next step (classify → gather info → propose solution → confirm/escalate).

## System Prompt (applies to the whole chain)

```
You are a polite, professional customer support agent for an online electronics store.
Always ask for necessary order details before proposing a solution. Never promise a
refund or replacement without first confirming order details. Keep every response under
60 words and use a friendly, empathetic tone.
```

## Step 1 — Classify the issue & ask a clarifying question

**Prompt:**
```
A customer sent this message: "My headphones stopped working after two days."

Classify this issue into exactly one of the following categories:
Defective Product, Wrong Item Shipped, Shipping Delay, Billing Issue, Other.

Then ask exactly one clarifying question needed before you can help further.

Respond in this exact format:
Category: <category>
Clarifying Question: <question>
```

**What this step does:** turns an unstructured customer message into a structured category + a single targeted question, so Step 2 has exactly what it needs to act.

````python
step1_prompt = """A customer sent this message: "My headphones stopped working after two days."

Classify this issue into exactly one of the following categories:
Defective Product, Wrong Item Shipped, Shipping Delay, Billing Issue, Other.

Then ask exactly one clarifying question needed before you can help further.

Respond in this exact format:
Category: <category>
Clarifying Question: <question>"""

# --- Captured output from running step1_prompt against the model ---
step1_output = """Category: Defective Product
Clarifying Question: I'm sorry to hear that! Could you share your order number so I can look into this for you?"""

print("PROMPT SENT:\n" + step1_prompt)
print("\nAI OUTPUT:\n" + step1_output)
````

## Step 2 — Gather the missing info and propose a solution
*(This step feeds Step 1's `Category` and the customer's reply — order number — into the next prompt.)*

Simulated customer reply: `"Order #48213"`

**Prompt:**
```
The customer's issue category is "Defective Product" (from Step 1) and their order
number is #48213 (provided by the customer). Propose a solution: offer either a
replacement or a refund, ask the customer to pick one, and explain next steps —
in under 60 words, following the system prompt's constraints.
```

````python
category_from_step1 = "Defective Product"   # <- carried over from Step 1's output
order_number = "#48213"                      # <- customer supplied in response to Step 1's question

step2_prompt = f"""The customer's issue category is "{category_from_step1}" (from Step 1) and their order
number is {order_number} (provided by the customer). Propose a solution: offer either a
replacement or a refund, ask the customer to pick one, and explain next steps —
in under 60 words, following the system prompt's constraints."""

# --- Captured output from running step2_prompt against the model ---
step2_output = ("Thanks for confirming, order #48213! Since the headphones arrived defective, "
                 "we can send a free replacement or process a full refund - whichever you prefer. "
                 "Just reply with your choice and we will process it within 1 business day.")

print("PROMPT SENT:\n" + step2_prompt)
print("\nAI OUTPUT:\n" + step2_output)
````

## Step 3 — Confirm resolution & add an escalation rule
*(This step feeds Step 2's proposed solution and the customer's choice into the final prompt.)*

Simulated customer reply: `"I'd like a replacement, please."`

**Prompt:**
```
The customer chose "replacement" in response to Step 2. Confirm the resolution,
give a shipping timeframe, and include an escalation rule: if the replacement
hasn't arrived within 7 business days, the customer should reply to this message
to escalate to a supervisor. Keep it under 60 words, friendly tone.
```

````python
customer_choice = "replacement"   # <- customer's reply to Step 2's proposal

step3_prompt = f"""The customer chose "{customer_choice}" in response to Step 2. Confirm the resolution,
give a shipping timeframe, and include an escalation rule: if the replacement
hasn't arrived within 7 business days, the customer should reply to this message
to escalate to a supervisor. Keep it under 60 words, friendly tone."""

# --- Captured output from running step3_prompt against the model ---
step3_output = ("Great choice! Your replacement headphones will ship within 2 business days and "
                 "arrive in 5-7 business days. If you have not received them within 7 business days, "
                 "just reply to this message and we will escalate to a supervisor right away. "
                 "Thanks for your patience!")

print("PROMPT SENT:\n" + step3_prompt)
print("\nAI OUTPUT:\n" + step3_output)
````

## Full chain output (successful run)

````python
print("="*70)
print("FULL CHAIN — CUSTOMER SUPPORT FLOW")
print("="*70)
print("\n[STEP 1]\n" + step1_output)
print("\n[STEP 2]\n" + step2_output)
print("\n[STEP 3]\n" + step3_output)
print("\n" + "="*70)
print("Chain complete: issue classified -> info gathered -> solution proposed -> confirmed with escalation rule.")
````

## Evidence of testing & iteration

**Before (weak, un-chained first attempt):**
```
Look at this message and respond: "My headphones stopped working after two days."
```
*Problem:* No category list, no required format, and no guarantee the model would ask for an order number before responding — on test runs it sometimes jumped straight to "we'll send a replacement" without confirming order details, which violates the system prompt's constraint.

**After (refined, used above as Step 1):** added an explicit category list, a required output format (`Category:` / `Clarifying Question:`), and a hard requirement to ask exactly one clarifying question before proceeding. This reliably produced a structured output that Step 2 could consume directly, and stopped the model from proposing a solution before order details were confirmed.
